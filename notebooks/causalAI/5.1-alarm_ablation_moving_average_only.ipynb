# Moving-Average-Only Alarm Ablation

Raise an alarm when the eight-sample moving average of the anomaly score from full-predictor models of the endogenous variables exceeds the threshold. The consecutive-alarm requirement is disabled. As in TEP notebook 5.1, the baseline window and threshold are kept fixed so this comparison removes only persistence. Evaluation uses all faulty and healthy held-out test runs.

```python
from pathlib import Path
import sys


def find_project_root(start=Path.cwd()):
    start = start.resolve()
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").is_file() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the project root from the notebook working directory.")


PROJECT_ROOT = find_project_root()
sys.path.insert(0, str(PROJECT_ROOT))
from IPython.display import display
from carl.causalai_baselines import run_alarm_ablation

alarm_results = run_alarm_ablation("moving_average_only", PROJECT_ROOT)
display(alarm_results["settings"])
display(alarm_results["faulty_test_metrics"])
display(alarm_results["healthy_test_metrics"])
display(alarm_results["comparison"])
```
