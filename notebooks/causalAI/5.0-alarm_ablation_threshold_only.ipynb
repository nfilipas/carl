# Threshold-Only Alarm Ablation

Raise an alarm when the raw anomaly score from full-predictor models of the endogenous variables exceeds the threshold. Neither moving-average smoothing nor consecutive-alarm persistence is used. As in TEP notebook 5.0, keep the threshold at the baseline value selected by notebook 3.0. Evaluation uses all faulty and healthy held-out test runs.

```python
from pathlib import Path
import sys


def find_project_root(start=Path.cwd()):
    start = start.resolve()
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").is_file() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the project root from the notebook working directory.")


PROJECT_ROOT = find_project_root()
sys.path.insert(0, str(PROJECT_ROOT))
from IPython.display import display
from carl.causalai_baselines import run_alarm_ablation

alarm_results = run_alarm_ablation("threshold_only", PROJECT_ROOT)
display(alarm_results["settings"])
display(alarm_results["faulty_test_metrics"])
display(alarm_results["healthy_test_metrics"])
display(alarm_results["comparison"])
```
